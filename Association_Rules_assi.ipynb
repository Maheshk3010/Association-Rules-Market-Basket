```python
import pandas as pd

df = pd.read_excel("Online retail.xlsx", header=None)
df.columns = ["Transaction"]
df.head()


```

```python


# Convert each transaction string into list of items

transactions = df["Transaction"].apply(lambda x: x.split(","))
transactions.head()
```

```python


from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

basket = pd.DataFrame(te_array, columns=te.columns_)
basket.head()
                                                      # Convert transactions to basket (0/1) format
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

frequent_itemsets = apriori(
    basket,
    min_support=0.02,
    use_colnames=True
)

frequent_itemsets.head()
                                   # Generate frequent itemsets using Apriori
```

```python

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.2
)

rules.head()
                                   # Generate association rules using confidence
```

```python
rules.shape
                                   # Check rules size
```

```python
rules.head()
rules[['support','confidence','lift']].describe()
```
