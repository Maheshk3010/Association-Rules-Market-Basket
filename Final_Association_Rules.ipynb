```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
```

```python
df = pd.read_excel('Online retail.xlsx', header=None)

df.head()
```

```python
transactions = []

for i in range(len(df)):
    transactions.append(str(df.iloc[i, 0]).split(','))

transactions[:5]
```

```python
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)

df_encoded = pd.DataFrame(te_data, columns=te.columns_)

df_encoded.head()
```

```python
from mlxtend.frequent_patterns import apriori

frequent_itemsets = apriori(df_encoded, min_support=0.02, use_colnames=True)

frequent_itemsets.head()
```

 Interview Questions

1. What is Lift and why is it important?

Lift measures how strongly two items are associated with each other.
It is calculated as:

Lift = Confidence / Support of consequent

If Lift > 1 → items are positively related  
If Lift = 1 → no relation  
If Lift < 1 → negative relation  

It is important because it helps us find meaningful and useful rules.

2. What is Support and Confidence? How are they calculated?

Support:
Support tells how frequently an itemset appears in the dataset.

Support(A) = (Number of transactions containing A) / (Total transactions)

Confidence:
Confidence tells how often item B is purchased when item A is purchased.

Confidence(A → B) = Support(A ∪ B) / Support(A)

3. What are limitations of Association Rule Mining?

- It can generate too many rules which are difficult to understand  
- It requires proper threshold selection (support, confidence)  
- It is computationally expensive for large datasets  
- Sometimes it finds irrelevant or non-useful rules
